## Function that load the .zip file, extract the unique object ids, and append the path to each individual image in .npy format, and orders as diff, scie, template.


Later fix the cases where there is more than 3 files. I did not check the case where there are less than 3 files.

```python
def create_file_list_save_pickle(path_type_obj):
  file_list_total = []
  observations = dict()
  zip_path = os.path.join(path_save, path_type_obj)

  with zipfile.ZipFile(zip_path, 'r') as zipper:
      # List all files inside the zip
      for file_name in zipper.namelist():
        print(file_name)
        if file_name.endswith('.npy'):
          obs_num = file_name.split("/")[-2]
          observations.setdefault(obs_num, []).append(file_name)
  for i, (k, v) in enumerate(observations.items()):
      file_list_total.append(sorted(v))

  lenghts = np.array([len(x) for x in file_list_total])

  return file_list_total, lenghts
```

## For vs_zip/VS_cutouts.zip

```python
file_list_total_vs, lenghts_vs = create_file_list_save_pickle(zip_paths[0])
```


### fix the cases where there is more than 3 files. I did not check the case where there are less than 3 files

```python
with open(os.path.join(path_save, f"{zip_paths[0].split('/')[0]}/file_list_vs_cutouts.pkl"), 'wb') as handle:
    pickle.dump(file_list_total_vs, handle)
```
